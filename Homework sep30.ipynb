Bitwise Operator---> 
&,^,~,<<,>>

Bitwise &

```python
a = 5 #0101
b = 3 #0011

print(a & b)#0001
```

```python
a = 12 #1100
b = 10 #1010

print(a & b) # 1000
```

Bitwise OR 

```python
a = 4 #0101
b = 2 #0010

print(a | b)# 0110
```

```python
a = 5 #0101
b = 3 #0011

print(a | b)#0111
```

Bitwise XOR

```python
a = 5 #0101
b = 3 #0011

print(a ^ b)
```

```python
a = 14 #1110
b = 9  #1001

print(a ^ b)
```

Right Shift >>

```python
a = 3
b = 4

print(3 >> 4)
```

```python
a = 8
b = 2

print(a >> b)
```

Left shift <<

```python
a = 3
b = 2

print(a << b)
```

```python
a = 2
b = 1

print(a << b)
```
